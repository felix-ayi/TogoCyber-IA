```python
import sys
from pathlib import Path
ROOT = Path.cwd()
if not (ROOT / 'ml').is_dir():
    ROOT = ROOT.parent
if not (ROOT / 'ml').is_dir():
    raise RuntimeError('Ouvrez ce notebook depuis le dépôt TogoCyber-AI.')
sys.path.insert(0, str(ROOT))
from ml.network.preprocessing import load_network_dataset
from ml.network.explain import explain_network
from ml.phishing.preprocessing import load_phishing_dataset
from ml.phishing.explain import explain_phishing

cic = ROOT / 'data/raw/cic_ids2017'
unsw_test = ROOT / 'data/raw/unsw_nb15/test.csv'
phishing_candidates = [ROOT / 'data/raw/phishing.csv', ROOT / 'data/raw/phishing/Phishing_Email.csv']
phishing_path = next((path for path in phishing_candidates if path.is_file()), None)
if cic.is_dir() and any(cic.rglob('*.csv')):
    network_path, network_dataset = cic, 'cic'
elif unsw_test.is_file():
    network_path, network_dataset = unsw_test, 'unsw'
else:
    raise FileNotFoundError('Ajoutez des données CIC ou téléchargez le test UNSW de démonstration.')
if phishing_path is None:
    raise FileNotFoundError('Ajoutez un corpus phishing autorisé ou téléchargez le corpus public de démonstration.')
network_x, _ = load_network_dataset(network_path, network_dataset)
texts, _ = load_phishing_dataset(phishing_path)
network_example = network_x.iloc[0].to_dict()
phishing_example = texts.iloc[0]
print('SHAP réseau:', explain_network(network_example))
print('LIME phishing:', explain_phishing(phishing_example))
```
